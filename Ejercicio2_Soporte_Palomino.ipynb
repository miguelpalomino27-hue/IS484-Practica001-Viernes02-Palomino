## Ejercicio 2: Agente de soporte de tickets


### Ficha PEAS

| Elemento                    | Descripción                                                                                         |
| --------------------------- | --------------------------------------------------------------------------------------------------- |
| **Percepción (S)**          | Tiempo de espera en minutos, nivel de urgencia y si el cliente es premium.                          |
| **Acciones (A)**            | Atender normalmente, priorizar el ticket o escalarlo al soporte especializado.                      |
| **Entorno (E)**             | Sistema de soporte que recibe y gestiona tickets de atención al cliente.                            |
| **Objetivo**                | Atender los tickets según su urgencia, tiempo de espera y condición del cliente.                    |
| **Medida de desempeño (P)** | Reducir los tiempos de atención y priorizar correctamente los casos urgentes y de clientes premium. |


### Justificación de las reglas

Las reglas permiten priorizar los tickets considerando la urgencia, el tiempo de espera y la condición del cliente. Los casos de urgencia alta se escalan inmediatamente para reducir el tiempo de atención. Los tickets con espera prolongada pueden recibir prioridad, especialmente cuando el cliente es premium. De esta manera, el equipo de soporte puede organizar mejor la atención y responder primero a los casos que requieren mayor atención.


```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):

    if nivel_urgencia == "alta":
        accion = "Escalar inmediatamente"
        motivo = "El ticket tiene urgencia alta y requiere atención inmediata."

    elif nivel_urgencia == "media" and tiempo_espera_minutos >= 30:
        accion = "Priorizar ticket"
        motivo = "La urgencia es media y el tiempo de espera es igual o mayor a 30 minutos."

    elif nivel_urgencia == "baja" and cliente_premium and tiempo_espera_minutos >= 30:
        accion = "Priorizar ticket"
        motivo = "El cliente es premium y el ticket tiene una espera prolongada."

    else:
        accion = "Atender normalmente"
        motivo = "El ticket no cumple las condiciones para ser priorizado o escalado."

    return accion, motivo
```

### Simulación y pruebas


```python
# Prueba 1: urgencia alta
print(agente_soporte(10, "alta", False))

# Prueba 2: urgencia media con espera prolongada
print(agente_soporte(30, "media", False))

# Prueba 3: urgencia media con espera corta
print(agente_soporte(20, "media", False))

# Prueba 4: urgencia baja, cliente premium y espera prolongada
print(agente_soporte(40, "baja", True))

# Prueba 5: urgencia baja, cliente premium y espera corta
print(agente_soporte(20, "baja", True))

# Prueba 6: urgencia baja, cliente no premium
print(agente_soporte(40, "baja", False))
```

```python
# Prueba 7: urgencia alta y cliente premium con espera prolongada
print(agente_soporte(60, "alta", True))

# Prueba 8: urgencia media, cliente premium y espera prolongada
print(agente_soporte(45, "media", True))
```
